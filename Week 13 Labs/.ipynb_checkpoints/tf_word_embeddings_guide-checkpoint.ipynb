# Word embeddings with Keras and TensorFlow

This notebook is based on the TensorFlow Text guide on word embeddings. It shows how to:

* Download and prepare the IMDb movie review sentiment dataset
* Vectorize raw text into integer sequences
* Train a simple sentiment classifier with an `Embedding` layer
* Export the learned word embeddings for visualization in the Embedding Projector

Original guide: https://www.tensorflow.org/text/guide/word_embeddings


## Setup

```python
import io
import os
import re
import shutil
import string

import tensorflow as tf

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Embedding, GlobalAveragePooling1D
from tensorflow.keras.layers import TextVectorization

print("TensorFlow version:", tf.__version__)
```

<!-- ## Download the IMDb dataset -->

We will use the Large Movie Review Dataset (IMDb) to train a binary sentiment classifier. Reviews in the `pos` folder are positive and reviews in the `neg` folder are negative.


```python
url = "https://ai.stanford.edu/~amaas/data/sentiment/aclImdb_v1.tar.gz"

dataset = tf.keras.utils.get_file(
    "aclImdb_v1.tar.gz",
    url,
    untar=True,
    cache_dir=".",
    cache_subdir=""
)

dataset_dir = os.path.join(os.path.dirname(dataset), "aclImdb")
print("Dataset directory:", dataset_dir)
print("Top-level contents:", os.listdir(dataset_dir))

train_dir = os.path.join(dataset_dir, "train")
print("Train directory contents:", os.listdir(train_dir))
```

The `train` directory also contains additional folders that we do not need. Remove the `unsup` directory and create `tf.data.Dataset` objects for training and validation using `text_dataset_from_directory`.


```python
# Remove the 'unsup' directory if it exists
remove_dir = os.path.join(train_dir, "unsup")
if os.path.exists(remove_dir):
    shutil.rmtree(remove_dir)
    print("Removed directory:", remove_dir)

batch_size = 1024
seed = 123

train_ds = tf.keras.utils.text_dataset_from_directory(
    train_dir,
    batch_size=batch_size,
    validation_split=0.2,
    subset="training",
    seed=seed,
)

val_ds = tf.keras.utils.text_dataset_from_directory(
    train_dir,
    batch_size=batch_size,
    validation_split=0.2,
    subset="validation",
    seed=seed,
)

class_names = train_ds.class_names
print("Class names:", class_names)
```

Inspect a few example reviews and labels. Labels are `0` for negative and `1` for positive reviews.


```python
for text_batch, label_batch in train_ds.take(1):
    for i in range(5):
        print("Label:", int(label_batch[i].numpy()))
        print("Review snippet:")
        print(text_batch.numpy()[i][:500])
        print("-" * 80)
```

## Configure the dataset for performance

To improve input pipeline performance, cache the datasets in memory (if they fit) and prefetch batches while the model is training.


```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=AUTOTUNE)
```

## The `Embedding` layer

An `Embedding` layer maps integer word indices to dense vectors of real numbers. You can think of it as a trainable lookup table.


```python
# Example: embed a vocabulary of size 1,000 into 5 dimensions
example_embedding = tf.keras.layers.Embedding(input_dim=1000, output_dim=5)

# A single sequence of token IDs
sample_ids = tf.constant([1, 2, 3])
sample_embedding = example_embedding(sample_ids)
print("Sample embedding shape:", sample_embedding.shape)
print(sample_embedding.numpy())

# A batch of sequences
batch_ids = tf.constant([[0, 1, 2], [3, 4, 5]])
batch_embedding = example_embedding(batch_ids)
print("Batch embedding shape:", batch_embedding.shape)
```

## Text preprocessing with `TextVectorization`

Define a preprocessing function that:

* Converts text to lowercase
* Replaces HTML line breaks (`<br />`) with spaces
* Strips punctuation

Then build a `TextVectorization` layer that tokenizes the text and maps tokens to integer IDs.


```python
# Custom standardization function
def custom_standardization(input_data: tf.Tensor) -> tf.Tensor:
    lowercase = tf.strings.lower(input_data)
    stripped_html = tf.strings.regex_replace(lowercase, "<br />", " ")
    return tf.strings.regex_replace(
        stripped_html,
        "[%s]" % re.escape(string.punctuation),
        ""
    )

# Vocabulary size and sequence length
vocab_size = 10000
sequence_length = 100

# TextVectorization layer
vectorize_layer = TextVectorization(
    standardize=custom_standardization,
    max_tokens=vocab_size,
    output_mode="int",
    output_sequence_length=sequence_length,
)

# Build the vocabulary on the training text
text_only_ds = train_ds.map(lambda x, y: x)
vectorize_layer.adapt(text_only_ds)

# Test the vectorizer
for text_batch, _ in train_ds.take(1):
    example_text = text_batch[0]
    print("Original review:")
    print(example_text.numpy()[:500])
    print("\nVectorized:")
    print(vectorize_layer(example_text)[0][:20])
    break
```

## Build a sentiment classification model

The model is a simple "continuous bag-of-words" style network:

1. `TextVectorization` turns raw strings into token IDs.
2. `Embedding` looks up a dense vector for each token.
3. `GlobalAveragePooling1D` averages embeddings across the time dimension.
4. Two dense layers perform binary sentiment classification.


```python
embedding_dim = 16

model = Sequential(
    [
        vectorize_layer,
        Embedding(vocab_size, embedding_dim, name="embedding"),
        GlobalAveragePooling1D(),
        Dense(16, activation="relu"),
        Dense(1),  # logits for binary classification
    ]
)

model.summary()
```

## Compile and train the model

Use the Adam optimizer and binary cross-entropy loss (with `from_logits=True` because the final layer outputs raw logits). Optionally, you can log to TensorBoard.


```python
tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir="logs")

model.compile(
    optimizer="adam",
    loss=tf.keras.losses.BinaryCrossentropy(from_logits=True),
    metrics=["accuracy"],
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=[tensorboard_callback],
)
```

You can view the training curves in TensorBoard (works in environments that support notebook magics such as Jupyter/Colab).

## Retrieve and save the trained word embeddings

The learned word embeddings are the weights of the `Embedding` layer. Save them, together with the vocabulary, in TSV files that can be uploaded to the Embedding Projector (https://projector.tensorflow.org).

```python
# Get the embedding weights
embedding_weights = model.get_layer("embedding").get_weights()[0]
print("Embedding matrix shape:", embedding_weights.shape)

# Get the vocabulary from the vectorization layer
vocab = vectorize_layer.get_vocabulary()
print("Vocabulary size (including padding):", len(vocab))

# Write vectors and metadata to disk
vectors_path = "vectors.tsv"
metadata_path = "metadata.tsv"

with io.open(vectors_path, "w", encoding="utf-8") as out_v, io.open(
    metadata_path, "w", encoding="utf-8"
) as out_m:
    for index, word in enumerate(vocab):
        # Skip the padding token at index 0
        if index == 0:
            continue
        vec = embedding_weights[index]
        out_v.write("\t".join([str(x) for x in vec]) + "\n")
        out_m.write(word + "\n")

print("Saved embedding vectors to:", vectors_path)
print("Saved embedding metadata to:", metadata_path)
```

If you run this notebook in Google Colab, you can trigger downloads for the TSV files using the helper below (wrapped in a `try/except` so that it is safe to run outside Colab).

```python
try:
    from google.colab import files

    files.download("vectors.tsv")
    files.download("metadata.tsv")
except Exception as e:
    print("Download helper is only available in some environments:", e)
```

## Visualize the embeddings

1. Open https://projector.tensorflow.org in your browser.
2. Click **"Load data"**.
3. Upload `vectors.tsv` as the **Tensor TSV file** and `metadata.tsv` as the **Metadata TSV file**.
4. Explore the embedding space by searching for words and inspecting their nearest neighbors.

You can experiment with different model architectures (for example, removing the hidden dense layer) and retrain to see how the learned embeddings change.
